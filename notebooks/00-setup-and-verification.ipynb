```python
import sys, subprocess
from pathlib import Path
from componergy.paths import ROOT, FIGURES_DIR
def run(script, *args):
    """Run a pipeline script exactly as on the command line (from the repository root)."""
    r = subprocess.run([sys.executable, '-m', f'scripts.{script}', *args], cwd=ROOT)
    print('exit code', r.returncode)
    return r.returncode

run('run_all', '--dry-run')
```
