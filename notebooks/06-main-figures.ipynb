```python
import os, re, sys, subprocess
from pathlib import Path
from componergy.paths import ROOT, FIGURES_DIR
NOISE = re.compile(r"findfont: Generic family|All-NaN slice encountered|return fnb\._ureduce")
def run(script, *args):
    """Run a pipeline script exactly as on the command line (from the repository root)."""
    env = {**os.environ, 'PYTHONUNBUFFERED': '1'}
    with subprocess.Popen([sys.executable, '-m', f'scripts.{script}', *args], cwd=ROOT, env=env,
                          stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True) as p:
        for line in p.stdout:
            if not NOISE.search(line):
                print(line, end='')
    print('exit code', p.returncode)
    return p.returncode

for n in ('01', '02', '03', '04', '05', '06', '07'):
    print(f'--- figure {n} ---'); run(f'run_figure_{n}')
```
